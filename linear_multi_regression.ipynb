```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
dataset=pd.read_csv("multi.csv")
```

```python
dataset
```

```python
m_height=dataset.height.mean()             ## fill nana values
```

```python
dataset.height=dataset.height.fillna(m_height)
```

```python
mult_reg=LinearRegression()
```

```python
mult_reg.fit(dataset[["AGE","height","weight"]],dataset[["PREMIUM"]])
```

```python
mult_reg.intercept_
```
