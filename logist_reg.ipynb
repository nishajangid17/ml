```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("logist_reg.csv")
```

```python
dataset
```

```python
dataset["insurance"].replace({"no":0,"yes":1},inplace=True)
```

```python
dataset
```

```python
plt.scatter(x="age", y="insurance",data=dataset)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=dataset[["age"]]
y=dataset[["insurance"]]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.score(x_test,y_test)*100
```

```python
lr.predict(x_test)
```

```python
lr.predict([[90]])
```

```python
lr.predict([[10]])
```

# predict line


```python
sns.scatterplot(x="age", y="insurance",data=dataset)
sns.lineplot(x="age",y=lr.predict(x),data=dataset,color="red")
plt.show()
```
